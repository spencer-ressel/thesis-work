```python
def set_plot_mode(mode):

    if mode == "publication":
        plt.rcParams.update({
            "figure.figsize": (3.5, 3),
            "font.size": 8,
            "axes.titlesize": 9,
            "axes.labelsize": 8,
            "lines.linewidth": 1.0,
        })

    elif mode == "slides":
        plt.rcParams.update({
            "figure.figsize": (8, 6),
            "font.size": 14,
            "axes.titlesize": 20,
            "axes.labelsize": 18,
            "lines.linewidth": 2.5,
        })

def get_figsize(mode, layout):

    FIG_SIZES = {
        "publication": {
            "onecol": (3.2, 2.6),
            "twocol": (5.5, 3.5),
            "full": (6.5, 5.0),
            "square": (3.2, 3.2),
        },
        "slides": {
            "onecol": (6, 4),
            "twocol": (10, 5),
            "full": (12, 7),
            "square": (6, 6),
        },
    }

    return FIG_SIZES[mode][layout]
```
