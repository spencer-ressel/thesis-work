# Convert model to pressure levels

```python
from geocat.comp.interpolation import interp_hybrid_to_pressure
import xarray as xr
import numpy as np
plevs = 100*np.array([  25,   50,   75,  100,  125,  150,  175,  200,  225,  250,  275,  300,
        325,  350,  375,  400,  425,  450,  475,  500,  525,  550,  575,  600,
        625,  650,  675,  700,  725,  750,  775,  800,  825,  850,  875,  900,
        925,  950,  975, 1000])


data_directory = "/glade/campaign/univ/uwas0152/post_processed_data/"

import glob
files_list = glob.glob(f"/glade/campaign/univ/uwas0114/SST_AQP3_Qobs_27_-4K_3h_10y/atm/hist/*h1*.nc")
non_timed_data = xr.open_dataset(files_list[0]).isel(time=0, drop=True)

experiments_list = ['-4K', '0K', '4K']
for variable in ['CLDLIQ', 'CLOUD']:
    print(f"Variable: {variable}")
    print(f"{'-'*40}")
    for experiment in experiments_list:
        print(f"{f'{experiment}...':<{40-1}}", end='')
        surface_pressure = xr.open_dataset(f"{data_directory}/{experiment}/daily_model-level_data/SST_AQP3_Qobs_27_{experiment}_1D_PS.nc")['PS']
        model_level_data = xr.open_dataset(
            f"{data_directory}/{experiment}/daily_model-level_data/SST_AQP3_Qobs_27_{experiment}_1D_{variable}.nc"
        )[variable]

        pressure_level_data = interp_hybrid_to_pressure(
            model_level_data,
            surface_pressure,
            non_timed_data['hyam'],
            non_timed_data['hybm'],
            p0=100000.0,
            new_levels=plevs
        ).assign_coords({'plev':plevs//100})

        pressure_level_data.to_netcdf(f"{data_directory}/{experiment}/daily_pressure-level_data/SST_AQP3_Qobs_27_{experiment}_1D_20y_{variable}.nc")

        print(rf"{'✔':>1}")

print(f"{'='*40}")
print("Finished")
```
